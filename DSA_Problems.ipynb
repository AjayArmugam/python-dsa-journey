### Problem 1: Even or Odd

```python
number = int(input("enter a number:"))

if number % 2 == 0:
  print(f"The {number} is Even")
else:
  print(f"The {number} is Odd")
```

## Problem 2 – Positive, Negative or Zero

```python
number = int(input("enter a number:"))

if number < 0:
  print(f"The {number} is negative")
elif number > 0:
  print(f"The {number} is positive")

else:
  print(f"The {number} is Zero")
```

## Problem 3 – Find the Largest of Two Numbers

```python
num1 = int(input("enter the first number:"))
num2 = int(input("enter the second number:"))

if num1 > num2:
  print(f"{num1} is Greater than {num2}")

elif num1 < num2:
  print(f"{num2} is Greater than {num1}")
else:
  print(f"1{num1} is Equal to the {num2}")
```

## Nested if

```python
num1 = int(input("enter the first number:"))
num2 = int(input("enter the second number:"))

if num1 >= num2:
  if num1 == num2:
    print(f"{num1} is Equal to {num2}")
  else:
    print(f"{num1} is Greater than {num2}")
else:
  print(f"{num2} is Greater than {num1}")
```

### Problem 4 – Find the Largest of Three Numbers

```python
num1 = int(input("enter the first number:"))
num2 = int(input("enter the second number:"))
num3 = int(input('enter the third number:'))

if num1 == num2 and num2 == num3:
  print("All three numbers are equal")

elif num1 >= num2 and num1 > num3:
  print(f"{num1} is Greater")

elif num2 >= num1 and num2 > num3:
  print(f"{num2} is Greater")

else:
  print(f"{num3} is Greater")
```

## Problem 5 – Check if a Number is Divisible by 5 and 3

```python
num = int(input("Enter a Number:"))

if num % 5 == 0 and num % 3 == 0:
  print(f"{num} Divisible by both 5 and 3")
elif num % 5 == 0:
  print(f"{num} Divisible by 5 Only")
elif num % 3 == 0:
  print(f"{num} Divisible by 3 Only")
else:
  print(f"{num} Not Divisible by 5 and 3")
```
